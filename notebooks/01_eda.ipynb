```python
import pandas as pd

df = pd.read_csv("../data/raw/ewallet_reviews.csv")
df.head()
```

```python
df.info()
df['app'].value_counts()
df['score'].value_counts().sort_index()
```

```python
# Buang skor 3 (netral), buat label sentimen
df = df[df['score'] != 3].copy()
df['sentiment'] = df['score'].apply(lambda x: 'positive' if x >= 4 else 'negative')

df['sentiment'].value_counts()
```

```python
df['content_length'] = df['content'].str.len()
df['content_length'].describe()
```

```python
# Lihat contoh review positif dan negatif
print("=== Contoh POSITIF ===")
print(df[df['sentiment']=='positive']['content'].sample(5).to_list())

print("\n=== Contoh NEGATIF ===")
print(df[df['sentiment']=='negative']['content'].sample(5).to_list())
```
